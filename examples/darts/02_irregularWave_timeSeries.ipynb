```python
import os
import pandas as pd
import matplotlib.pyplot as plt
```

Import and Interpolate the data to a given timestep

```python
waveData = 'waveData.csv'
waveData = pd.read_csv(waveData)

waveData = waveData.set_index('Time')

# Choose your desired frequency (e.g., 0.5s)
uniform_index = pd.date_range(start=waveData.index[0],
                               end=waveData.index[-1],
                               freq='500ms')
                               

# Reindex and interpolate
waveData = waveData.reindex(uniform_index)
waveData = waveData.interpolate(method='linear')  # fill NaNs
waveData = waveData.reset_index().rename(columns={'index': 'Time'})
```
